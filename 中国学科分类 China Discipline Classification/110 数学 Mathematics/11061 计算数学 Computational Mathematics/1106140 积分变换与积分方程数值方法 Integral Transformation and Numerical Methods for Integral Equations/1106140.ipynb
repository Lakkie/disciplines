## 什么是积分变换？ 
What is Integral Transformation？

定义：积分变换是一种通过‌积分运算‌将某类函数从‌原定义域‌（如时间域、空间域）映射到‌变换域‌（如频率域、复平面域）的数学工具。其核心思想是利用特定的‌核函数（Kernel）‌与原函数进行积分运算，生成新的函数，从而简化原问题的分析或求解。
Definition: Integral transformation is a mathematical tool that maps a certain type of function from the original domain (such as time domain, space domain) to the transformation domain (such as frequency domain, complex plane domain) through integral operation. The core idea is to use a specific kernel function to integrate with the original function, generate new functions, and simplify the analysis or solution of the original problem.

积分变换是一类通过积分运算将函数从原域（如时域）映射到变换域（如频域或复频域）的数学工具，广泛应用于工程数学、物理学及信号处理等领域‌。
Integral transformation is a mathematical tool that maps functions from the original domain (such as the time domain) to the transformation domain (such as the frequency domain or complex frequency domain) through integral operations. It is widely used in fields such as engineering mathematics, physics, and signal processing.

## 积分变换的一般形式
The general form of integral transformation

积分变换的数学表达式   
设原函数为 f(t)，积分变换的一般形式为：
\begin{equation}
F(s) = \int_{a}^{b} K(s, t) \, f(t) \, dt
\end{equation}
其中：
K(s,t) 是‌核函数‌，决定了变换的性质和映射规则。   
a 和 b 是积分区间（通常为 −∞ 到 +∞，或 0 到 +∞）。   
F(s) 是原函数 f(t) 在变换域中的像函数（Image Function）。   
通过选择不同的核函数和积分区间，可定义不同类型的积分变换。   

## 傅里叶变换
Fourier Transform

（连续）傅里叶变换（Continuous Fourier Transform）是一种将 ‌时域信号‌ f(t) 转换为 ‌频域信号‌ F(ω) 的积分变换，揭示信号中不同频率成分的分布。
\begin{equation}
\mathcal{F}\{f(t)\}(\omega) = \int_{-\infty}^{\infty} f(t) \, e^{-i \omega t} \, dt
\end{equation}
‌输入‌：时间（或空间）函数 f(t)。   
‌输出‌：频率函数 F(ω)，表示信号在角频率 ω 上的复振幅。   
核函数‌：（复指数函数，对应正弦/余弦基函数）
\begin{equation}
\mathcal{K}(\omega, t) = e^{-i \omega t}
\end{equation}   
‌积分区间‌：(−∞,∞)   
逆变换：
\begin{equation}
\mathcal f(t)\ = \frac{1}{2\pi} \int_{-\infty}^{\infty} F(\omega)  \, e^{i \omega t} \, d\omega
\end{equation}
应用‌：将时域信号 f(t) 转换为频域 F(ω)，广泛用于信号分析、量子力学和热传导方程求解。

Laplace Transform

\begin{equation}
\mathcal{L}\{f(t)\}(s) = \int_{0}^{\infty} f(t) \, e^{-s t} \, dt

\begin{equation}

\end{equation}
