## 1.1. 线性规划问题的变体 Variants of the linear programming problem 

In this seition, we pose the linear programming problem, discuss a few special forms that it takes, and establsh some standard rotation that we will be using. Rather than starting abstractly. we first state a concrete example, which is meant to fecilitate understanding of the formal definition that will follow. The example we give is devoid of any interpretation. Later on, in Section 1.2, we will have ample opportunity to develp examples that arise in practical settings.  

在本节中，我们将提出线性规划问题，讨论其一些特殊形式，并建立一些我们将要使用的标准旋转。我们不会从抽象的角度开始，而是首先陈述一个具体的例子，旨在帮助理解接下来的正式定义。我们给出的例子不包含任何解释。稍后，在1.2节中，我们将有充足的机会来开发实际场景中出现的例子。

Example 1.1 The following is a linear programming problem:  
minimize  
$$2x_1 - x_2 + 4x_3$$
subject to  
$$x_1 + x_2 - x_4 \leq 2$$
$$3x_2 - x_3 = 5$$
$$x_3 - x_4 \geq 3$$
$$x_1 \geq 0$$
$$x_3 \leq 0$$

目标函数 求最小值 minimize  
$$2x_1 + (-1)x_2 + 4x_3 + 0x_4 $$
约束条件 subject to  
$$1x_1 + 1x_2 + 0x_3 + (-1)x_4 \leq 2$$
$$0x_1 + 3x_2 + (-1)x_3 + 0x_4 = 5$$
$$0x_1 + 0x_2 + 1x_3 + (-1)x_4 \geq 3$$
$$1x_1 + 0x_2 + 0x_3 + 0x_4 \geq 0$$
$$0x_1 + 0x_2 + 1x_3 + 0x_4 \leq 0$$

Here $x_1$, $x_2$, $x_3$, and $x_4$ are variables whose values are to be closen to minimize the linear cost function $2x_1 - x_2 + 4x_3$, subject to a set of linear equality and inequality constraints. Some of those constraints, such as $x_1 \geq 0$  and $x_3 \leq 0$, amount to simple restrictions on the sign of certain variables. The remining constraints are of the form $a'x \leq b$, $a'x = 0$ or $a'x \geq b$, where $a = (a_1, a_2, a_3, a_4)$ is a given vector$^1$, $x = (x_1, x_2, x_3, x_4)$ is the vector of decision variables, $a'x$ is their inner product $\sum_{i=1}^{4} a_ix_i$, and b is a given scalar. For example, in the first constraint, we have $a = (1, 1, 0, 1)$ and $b = 2$.  

这里，$x_1$, $x_2$, $x_3$ 和 $x_4$ 是变量，它们的值需要被逼近最小化线性成本函数 $2x_1 - x_2 + 4x_3$，并受到一组线性等式和不等式约束。其中一些约束，例如 $x_1 \geq 0$ 和 $x_3 \leq 0$，相当于对某些变量符号的简单限制。剩余约束的形式为 $a'x \leq b$、$a'x = 0$ 或 $a'x \geq b$，其中 $a = (a_1, a_2, a_3, a_4)$ 为给定向量$^1$，$x = (x_1, x_2, x_3, x_4)$ 为决策变量向量，$a'x$ 为它们的内积 $\sum_{i=1}^{4} a_ix_i$，b 为给定标量。例如，在第一个约束中，我们有 $a = (1, 1, 0, 1)$ 和 $b = 2$。

$^1$As discussed further in Section 1.5, all vectors are assumed to becolummn vecors, and are treated as such in matrix-vecor products. Row vectors are indicated as transposes of (column) vectors. However, whenever we refer to a vector x inside the text, we use the more economical notation $x = (x_1, ..., x_n)$, even though x is a column vector. The reader who is unfamiliar with our notaion may wish to consult Section 1.5 before continuing.  

$^1$如 1.5 节进一步讨论的那样，所有向量都被假定为列向量，并在矩阵向量积中如此处理。行向量表示为（列）向量的转置。然而在文中每当我们提到向量 x 时，我们都使用更简洁的符号 $x = (x_1, ..., x_n)$，即使 x 是列向量。不熟悉我们符号的读者可能需要在继续阅读之前先查阅 1.5 节。

We now generalize. In a general linear programming problem. we are given a cost vector $c = (c_1, ..., c_n)$ aud we seek to minimize a Linear cost
function $c'x = \sum_{i=1}^{n} c_ix_i$ over all n-dimensional vectors $x = (x_1, ..., x_n)$, subject to a set of linear equality and inequality constraints. In particular, let $M_1$, $M_1$, $M_3$ be some finite index sets, and suppose that for every i in any one of these sets, we are given an n-dimensional vector $a_i$ and a scalar $b_i$, that will be used to form the ith constraint. Let also $N_1$ and $N_2$ be subsets of {1, ..., n} that indicate which variables $x_j$ are constrained to be nonnegatiive or nonpositive, respectively. We then consider the problem  

现在我们进行推广。在一个一般的线性规划问题中，给定一个成本向量 $c = (c_1, ..., c_n)$，我们试图最小化一个线性成本函数 $c'x = \sum_{i=1}^{n} c_ix_i$，该函数在所有 n 维向量 $x = (x_1, ..., x_n)$ 上，并满足一组线性等式和不等式约束。具体而言，设 $M_1$、$M_2$、$M_3$ 为一些有限索引集，假设对于这些集合中任何一个集合中的每个 i，我们都有一个 n 维向量 $a_i$ 和一个标量 $b_i$，它们将用于构成第 i 个约束。同时，设 $N_1$ 和 $N_2$ 为 {1, ..., n} 的子集，分别指示哪些变量 $x_j$ 被约束为非负或非正。然后我们考虑这个问题  

目标函数 求最小值 minimize $$c'x $$  
约束条件 subject to  
$$a_{i}^{'} \geq b_i, \ \ \ i \in M1,$$
$$a_{i}^{'} \leq b_i, \ \ \ i \in M2,$$
$$a_{i}^{'} = b_i, \ \ \ i \in M3,$$
$$x_j \geq 0, \ \ \ j \in N1,$$
$$x_j \leq 0, \ \ \ j \in N2.$$

The variables $x_1, ..., x_n$ are called decision variables, and a vector x satisfying all of the constraints is called a feasible solution or feasible vector. The set of all feasible solutions is called the feasible set or feasible region. If j is in neither $N_1$ nor $N_2$, there are no restrictions on the sign of $x_j$, in which case we say that $x_j$ is a free or unrestricted variable. The function $c^{\prime}x$ is called the objective function or cost function. A feasible solution $x^*$ that minimizes the objective function (that is, $c^{\prime}x^* \leq c^{\prime}x$, for all feasible x) is called an optimal feasible solution or, simply, an optimal solution. The valve of $c^{\prime}x^*$ is then called the optimal cost. On the other hand, if for every real number K we can find a feasible solution x whose cost is less than K, we say that the optima. cost is $-\infty$ or that the cost is unbourded below. (Sometimes, we will abuse terminology and say that the problem is unbounded.) We finally note that there is no need to study maximization problems separately, because maximizing $c^{\prime}x$ is equivalent to minimizing the linear cost function $-c^{\prime}x$.  

变量 $x_1, ..., x_n$ 称为决策变量，满足所有约束条件的向量 x 称为可行解或可行向量。所有可行解的集合称为可行集或可行域。如果 j 既不在 $N_1$ 也不在 $N_2$ 中，则 $x_j$ 的符号不受限制，在这种情况下，我们称 $x_j$ 为自由变量或无限制变量。函数 $c^{\prime}x$ 称为目标函数或成本函数。最小化目标函数的可行解 $x^*$（即，对于所有可行 x，$c^{\prime}x^* \leq c^{\prime}x$）称为最优可行解，或简称为最优解。$c^{\prime}x^*$ 的值称为最优成本。另一方面，如果对于每个实数 K，我们都能找到一个成本小于 K 的可行解 x，我们称其为最优解。成本为 $-\infty$，或者成本在下方无界。（有时，我们会滥用术语，称该问题是无界的。）最后，我们注意到，无需单独研究最大化问题，因为最大化 $c^{\prime}x$ 等价于最小化线性成本函数 $-c^{\prime}x$。
