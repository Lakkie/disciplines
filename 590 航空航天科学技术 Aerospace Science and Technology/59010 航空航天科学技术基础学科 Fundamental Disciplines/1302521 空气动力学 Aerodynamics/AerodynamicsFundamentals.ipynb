# 1 空气动力学绪论 Introduction to Aerodynamics

## 1.1 空气动力学研究任务 Aerodynamics Research Tasks

### 1.1.1 确定气动力与力矩 Determining aerodynamic forces and moments

研究飞行器在不同飞行状态（如速度、高度、姿态）下所受的升力、阻力、侧向力以及滚转、偏航和俯仰力矩，是空气动力学的基本任务。这些参数直接影响飞行性能、稳定性和操纵性。

气动力 Aerodynamic forces  
涵盖升力（lift）、阻力（drag）和侧向力（side force）

气动力矩 Aerodynamic moments  
包括俯仰力矩（pitching moment）、滚转力矩（rolling moment）和偏航力矩（yawing moment）。

### 1.1.2 分析流场结构与流动现象 Analyzing flow field structure and flow phenomena

通过研究物体周围的气流速度场、压力场、温度场分布，识别关键流动现象，如：  
激波与边界层相互作用  
气流分离与旋涡生成  
湍流与非定常流动  
动态失速与气动弹性效应  
这些现象对飞行安全和效率有重大影响。

流场结构 Flow field structure  
指流体运动区域内速度场、压强场、涡量场等物理量的空间组织形态，常用于揭示流动的稳定性、分离区、涡旋结构及干扰机制。例如：  
在压气机转子通道中，不同周向位置下存在四种典型的流场结构类型。  
横向喷流引起的三维复杂干扰流场结构可通过PIV技术进行测量与分析。  

流动现象 Flow phenomena  
涵盖流体在特定条件下的宏观行为表现，如层流（laminar flow）、湍流（turbulent flow）、流动分离（flow separation）、旋涡形成（vortex formation）等。例如：  
当雷诺数较低时，流体呈层流状态；随雷诺数增大，逐步过渡为湍流。  
在管道或多相流系统中，气液两相可呈现分层流（stratified flow）、环状流等典型流动现象。

### 1.1.3 开展气动热与高温气体效应研究 Conducting research on aerodynamic heating and high-temperature gas effects

在高速（尤其是高超声速）飞行条件下，空气被剧烈压缩产生高温，引发气体分子离解、电离等非平衡化学反应。研究高温气体的流动特性、热传导机制及对材料的烧蚀作用，是航天器再入大气层设计的关键。

气动热 Aerodynamic heating  
指高速飞行时气流动能通过压缩与黏性耗散转化为热能，导致飞行器表面温度急剧升高的物理过程。该现象在马赫数超过5时尤为显著，可使表面温度达2200K以上，构成“热障”挑战。
相关术语：aeroheating（气动热的缩略形式）、thermal loading（热载荷）、heat flux（热流密度）

高温气体效应 High-temperature gas effects  
涵盖气体在极端温度下偏离理想气体假设的多种物理化学行为，包括比热容变化、分子振动激发、离解与电离等真实气体效应。该效应直接影响气动力/热预测精度，是气动热力学（aerothermodynamics）研究的关键内容。
相关术语：real gas effects、chemical nonequilibrium、radiative heating、plasma sheath（等离子鞘套）

### 1.1.4 支撑气动外形设计与优化 Supporting aerodynamic configuration design and optimization

通过理论计算、风洞实验和数值模拟（CFD），评估不同构型的气动性能，优化飞行器外形以实现低阻力、高升力、良好稳定性和隐身性能。例如，F-22战斗机的隐身能力70%依赖于气动外形设计。

气动外形设计 Aerodynamic configuration design  
指飞行器总体气动布局的构建，包括机翼、机身、尾翼、进气道等部件的几何形状与空间组合方式。典型设计如：  
后掠翼布局（swept-wing configuration）用于减小跨声速阻力；  
升力体（lifting body）与翼身融合（blended wing body, BWB）结构提升升阻比；  
高超声速飞行器采用尖锐前缘与小展弦比布局以控制激波结构。

气动优化 Aerodynamic optimization  
指在给定约束条件下，通过参数化建模、敏感性分析与智能算法（如遗传算法、代理模型）搜索最优或近优气动外形。常见方法包括：  
基于CFD的梯度优化（adjoint-based optimization）；  
多目标优化（multi-objective optimization）兼顾升阻比与热防护性能；  
拓扑优化（topology optimization）探索新型气动构型。

### 1.1.5 发展实验技术与测试方法 Developing experimental techniques and testing methods

建设风洞、激波管、磁悬浮轨道等地面模拟设施，发展粒子图像测速（PIV）、压力敏感漆（PSP）、红外热成像等先进测量技术，确保实验数据的准确性和可重复性。

### 1.1.6 推动多学科融合研究 Promoting interdisciplinary research

空气动力学已与结构力学、控制理论、声学、热力学等深度融合，形成：  
气动/结构/控制一体化设计  
气动声学（降低飞行噪声）  
气动热力学（热防护系统）  
流固耦合分析等前沿方向。
