# Introduction 引言

# 1 What is a graph? 什么是图？

graph / general graph 图

一个图是一组顶点以及顶点如何连接的表示。

vertice / node 顶点

edge 边

multiple edges 多重边

一对顶点之间存在多条边。

degree 度

顶点的度是以此顶点为端点的边的数量。

simple graphs 简单图

没有环，且没有多重边的图

loop 环

directed graphs / digraphs 有向图

walk 游走

A walk is a 'way of getting from one vertex to another', and consists of a sequence of edges, one following after another. 
一个顶点到达另一个顶点的方式。

path 路径

A walk in which no vertex appears more than once is called a path.
游走中的每个顶点只出现一次。

cycle 环

游走从起点出发又回到该起点。

Eulerian graphs 欧拉图

图中包含恰好经过每条边一次的游走，并且该游走在初始顶点结束。

Hamiltonian graphs 哈密尔顿图

图中包含恰好经过每个顶点一次的游走，并且该游走在初始顶点结束。

connected graph 连通图

任何两个顶点都可以由一条路径相连。

disconnected graph 非连通图

一个图由多个部分组成，其中每个部分可能是一个连通图。

Isomorphism 同构性

Connectedness 连通性

Adjacency 邻接性

Subgraphs 子图

Matrix representations 矩阵表示

adjacency matrix 邻接矩阵

incidence matrix 关联矩阵
